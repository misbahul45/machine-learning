# Data Understanding

## Data Generating Process (DGP)
The Data Generating Process (DGP) is the true, underlying mechanism that produces the data we observe. It is unknown, because we never see it directly; we only see the samples it generates. For example, customer purchases might be generated by income, preferences, marketing exposure, and random noise. In machine learning, the DGP is the "ground truth" we are trying to learn: every dataset is just a finite snapshot drawn from this hidden process. Because the data alone is incomplete and noisy, our goal is not to memorize the data, but to recover the structure of the process that created it, including its patterns, relationships, and sources of randomness.

## Approximating the DGP with a Probability Model

Since the true DGP is unknown, we approximate it with an assumed probability model: a mathematical guess about how the data is generated. This is where distributions come in, such as Normal for continuous measures (e.g., height, prices), Bernoulli for binary outcomes (e.g., click / no click), and Poisson for counts (e.g., number of visits). The model also specifies structure, meaning how inputs relate to outputs, for example *y = β₀ + β₁x + ε*. The statistical engine, which includes estimation methods like maximum likelihood, least squares, or Bayesian updating, then uses the observed data to fit the model's parameters. Performance is maximized when the assumed model matches the true DGP closely: estimates become unbiased, uncertainty is quantified correctly, and predictions generalize to new data. If the model is misspecified (wrong distribution, missing relationships), diagnostics like residual analysis and cross-validation reveal the mismatch, so we refine the model and iterate, step by step getting closer to the real process behind the data.

## Domain Context

Domain context is about how we approach data before using it: the point of view we take, the goal or purpose behind our analysis, and which domain or subdomain the data actually belongs to. Understanding data becomes much easier when we know what the data is a representation of, and what we should do with it based on our purpose, because the same dataset can mean different things depending on context: sales data viewed through a marketing lens answers campaign questions, while the same data viewed through a finance lens answers revenue questions. So, before analyzing any data, we should first identify its topic, meaning what real-world process the data represents and why we are looking at it, so that the analysis is guided by both the domain it comes from and the purpose we have for it.

## Key Concept
The Data Generating Process (DGP) is the true, unknown mechanism that produces the data we observe; we never see it directly, only the samples it generates. Since the DGP is hidden, we approximate it with a probability model, which is our mathematical guess about how the data was created, using distributions like Normal for continuous measures, Bernoulli for binary outcomes, and Poisson for counts. The statistical engine (maximum likelihood, least squares, Bayesian updating) fits this model's parameters using the observed data, and performance is maximized when our assumed model matches the true DGP closely. Before any of this, domain context shapes everything: the point of view we take, the purpose of our analysis, and which domain or subdomain the data belongs to. Knowing what the data represents and why we are looking at it makes analysis easier and more accurate, because the same data means different things in different contexts. Together, these concepts form a chain: domain context tells us why we look at the data, the DGP is the real process behind it, the dataset is what we observe, and the probability model is how we approximate that process to extract knowledge and make predictions.

## Implementation Focus: Documenting Data Provenance

This is the practical side of the DGP and domain context ideas: when working with real data, you can't observe the generating process directly, so you **document it** instead. Documenting the origin of the dataset means recording everything known about how the data was produced:

**1. Origin of the dataset**
Where did the data come from and why does it exist? Examples: a sales dataset generated by a POS system, a log dataset generated by server events, a medical dataset from hospital records. Knowing the origin tells you what real-world process (the DGP) left this trace.

**2. Sensor specifications**
For sensor or IoT data: what device measured it, what unit, what accuracy and range, how often it samples, and how it fails (drift, noise, missing readings). A temperature sensor with ±2°C error means your data's "noise" component is partly known before any analysis.

**3. Survey methodologies**
For survey data: who was asked, how were they sampled, what was the exact question wording, and who was excluded? This tells you about **selection bias**, a key part of the DGP: the data you see was not generated from the whole population, only from the sampled part.

**4. Database extraction logic**
For data pulled from systems: which tables were joined, which filters were applied, how were duplicates and missing rows handled, and when was the snapshot taken? Two analysts writing different SQL on the same database effectively create **two different DGPs**, which explains why their numbers never match.

**Why this matters:**

Every documented detail is a clue about the true DGP. The more you know about how the data was generated, the better you can choose a correct probability model, detect bias and noise, and avoid false conclusions. In one sentence: **since we never see the DGP directly, documenting provenance (origin, sensors, methods, extraction logic) is how we reconstruct it as much as possible before analysis begins.**
